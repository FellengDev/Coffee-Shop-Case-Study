```sql
SELECT *
FROM workspace.salesdb.bright_coffee_shop_analysis
LIMIT 10;

--this is our unique Identifier 
SELECT COUNT(`transaction_id`) AS total_transactions,
       COUNT (DISTINCT `transaction_id` ) AS disctinct_transactions
FROM workspace.salesdb.bright_coffee_shop_analysis

----transaction date
SELECT
   YEAR(transaction_date) AS year,
   QUARTER(transaction_date) AS quater_number,
   MONTH(transaction_date) AS month_number, 
   DATE_FORMAT(transaction_date, 'EEEE') AS day_name, 

   CASE
       WHEN DATE_FORMAT(transaction_date, 'EEEE') IN ('Saturday', 'Sunday') THEN 'Weekend'
       ELSE 'Weekday'
   END AS day_type
FROM workspace.salesdb.bright_coffee_shop_analysis



----all the information happened on the same day
SELECT
    MIN(DATE(transaction_time)) AS min_transaction,
    MAX(DATE(transaction_time)) AS max_transaction
FROM workspace.salesdb.bright_coffee_shop_analysis

----min is 6am and max is 20:59:32
SELECT
    MIN(DATE_FORMAT(transaction_time, 'HH:mm:ss')) AS min_transaction,
    MAX(DATE_FORMAT(transaction_time, 'HH:mm:ss')) AS max_transaction
FROM workspace.salesdb.bright_coffee_shop_analysis

---there are different store locations
SELECT DISTINCT store_location AS distinct_store
FROM workspace.salesdb.bright_coffee_shop_analysis

SELECT transaction_qty,
      SUM(transaction_qty) AS total_transactions
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY transaction_qty;


SELECT product_category
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_category; 


SELECT DISTINCT product_type
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_type;

SELECT product_detail
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_detail; 
------------------------------------------------------------------------------------------------------------------------------
----KPI's 
SELECT
  ROUND(SUM(transaction_qty * unit_price ), 2) AS total_revenue
FROM workspace.salesdb.bright_coffee_shop_analysis

SELECT
   SUM(transaction_qty) AS total_quantity_sold 
FROM workspace.salesdb.bright_coffee_shop_analysis

SELECT
   ROUND(AVG(unit_price), 2) AS average_unit_price
FROM workspace.salesdb.bright_coffee_shop_analysis

----Revenue by product_category
SELECT product_category,
  ROUND(SUM(transaction_qty * unit_price), 2 ) AS revenue
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_category
ORDER BY revenue DESC;

---Revenue by product_type
SELECT product_type,
      ROUND(SUM(transaction_qty * unit_price), 2) AS revenue_type
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_type
ORDER BY revenue_type DESC;

SELECT
     HOUR(transaction_time) AS hour_transaction_time,
     ROUND(SUM(transaction_qty * unit_price), 2) AS total_revenue
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY HOUR(transaction_time)
ORDER BY HOUR(transaction_time);

---Top selling products
SELECT product_detail,
      SUM(transaction_qty) AS units_sold
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_detail
ORDER BY units_sold DESC;

----Highest revenue products
SELECT product_detail,
     ROUND(SUM(transaction_qty * unit_price), 2) AS revenue2
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY product_detail
ORDER BY revenue2 DESC;

----Sales by store 
SELECT store_location,
    ROUND(SUM(transaction_qty * unit_price), 2) AS revenue3
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY store_location
ORDER BY revenue3 DESC;

---units sold by store 
SELECT store_location,
       SUM(transaction_qty)AS units_per_store
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY store_location
ORDER BY units_per_store DESC;

---Sales by the hour
SELECT
    HOUR(transaction_time) AS hour_of_day,
    ROUND(SUM(transaction_qty * unit_price), 2) AS revenue4
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY HOUR(transaction_time)
ORDER BY hour_of_day DESC;

---Transaction by the hour
SELECT
    HOUR(transaction_time) AS hour_of_day1,
    COUNT(transaction_id) AS total_transaction
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY HOUR(transaction_time)
ORDER BY hour_of_day1 DESC;
 
---Revenue by day of week
SELECT
   DATE_FORMAT(transaction_date, 'EEEE') AS day_name,
   ROUND(SUM(transaction_qty * unit_price), 2) AS revenue5
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY DATE_FORMAT(transaction_date, 'EEEE')
ORDER BY revenue5 DESC;

---Weekday vs weekend
SELECT
     CASE
        WHEN DATE_FORMAT(transaction_date, 'EEEE') IN ('Saturday', 'Sunday') THEN 'Weekend'
        ELSE 'Weekday'
    END AS day_type2,
      ROUND(SUM(transaction_qty * unit_price), 2) AS day_revenue
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY 
       CASE
          WHEN DATE_FORMAT(transaction_date, 'EEEE') IN ('Saturday', 'Sunday') THEN 'Weekend'
          ELSE 'Weekday'
    END;

---AVG unit price
SELECT 
   ROUND(AVG(unit_price), 2) AS avg_unit_price,
   MIN(unit_price) AS lowest_unit_price,
   MAX(unit_price) AS max_unitprice
   FROM workspace.salesdb.bright_coffee_shop_analysis

   ---Number of products sold 
SELECT
     COUNT(DISTINCT product_detail) AS total_products
FROM workspace.salesdb.bright_coffee_shop_analysis

----Revenue by month
SELECT
        MONTHNAME(transaction_time) AS monthly_transaction,
        ROUND(SUM(transaction_qty * unit_price), 2) AS total_monthly_sales 
FROM workspace.salesdb.bright_coffee_shop_analysis
GROUP BY MONTHNAME(transaction_time)
ORDER BY monthly_transaction;

WITH CTE AS (
SELECT *,
   transaction_qty * unit_price AS revenue,
   YEAR(transaction_time) AS year_transaction_time,
   MONTH(transaction_time) AS month_trancastion_time,
   QUARTER(transaction_time) AS quarter_transaction_time,
   HOUR(transaction_time) AS hour_transition_time,
   DATE_FORMAT(transaction_time, 'EEEE') AS date_transaction_time,
   CASE
      WHEN DATE_FORMAT(transaction_time, 'EEEE') IN ('Saturday', 'Sunday') THEN 'Weekend'
      ELSE 'Weekday'
    END AS day_type1,

    CASE
        WHEN HOUR(transaction_time) BETWEEN 6 AND 11 THEN 'Morning'
        WHEN HOUR(transaction_time) BETWEEN 12 AND 16 THEN 'Afternoon'
        WHEN HOUR(transaction_time) BETWEEN 17 AND 20 THEN 'Evening'
        ELSE 'Night'
    END AS time_band
FROM workspace.salesdb.bright_coffee_shop_analysis

)
```
